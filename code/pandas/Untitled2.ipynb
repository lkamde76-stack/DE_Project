```python
def groupfnc(df, idx, col):
    if 80<= df[col].loc[idx] <=90:
        return '80-90 range'
    elif 50<= df[col].loc[idx] <=60:
        return '50-60 range'
    else:
        return "Others"
```

```python
wg = df_wc.groupby(lambda idx: groupfnc(df_wc,idx,'temperature'))
wg
```

```python
for key , data in wg:
    print("Key", key)
    print("Data",data)
```
