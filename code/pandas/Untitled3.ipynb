```python
import pandas as pd
```

```python
df_sales = pd.read_excel("linechart.xlsx")
df_sales
```

```python
from matplotlib import pyplot as plt
plt.figure(figsize=(12,4))
plt.plot(df_sales["Quarter"],df_sales["Fridge"],df_sales["Dishwasher"])
plt.title("Product sales")
plt.xlabel("Financial Quoter ")
plt.ylabel("Revenue in $ ")
plt.show()
```

```python
total_sales = df_sales[["Washing Machine","Fridge","Dishwasher"]].sum()
total_sales
```

```python
total_sales.index
```

```python
plt.pie(total_sales, labels = total_sales.index,autopct="%1.2f%%")
plt.show()
```
